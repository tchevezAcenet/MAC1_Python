# JupyterLab Interface

The interface of JupyterLab includes a **menu bar** at the top, a **left sidebar**, and the **main working area**, which contains tabs of notebooks, consoles, terminals, documents and activities. The menu bar at the top of JupyterLab has the top-level menus that show the various actions available in JupyterLab.

The following menus are included in the top menu by default.


- **File:** Actions related to files and directories such as New, Open, Close, Save, etc. The File menu also includes the *Shut Down* action used to shut down the JupyterLab server.
- **Edit:** Actions related to editing documents and other activities such as Undo, Cut, Copy, Paste, etc.
- **View:** Actions that change the appearance of JupyterLab.
- **Run:** Actions for running code in different activities such as notebooks and code consoles (discussed below).
- **Kernel:** Actions for managing kernels. Kernels in Jupyter are processes that are responsible for running your code. You can think of the kernel as **the engine behind your notebook**: it works with the Python interpreter to translate your code into machine language, sends the results back to the Jupyter interface, and remembers what you’ve done.

  When you press **Shift + Enter** to run a cell, the kernel executes that code and gives you a result (a number, a list, a graph, etc.).

- **Tabs:** A list of the open documents and activities in the main work area.
- **Settings:** Common JupyterLab settings can be configured using this menu. There is also an *Advanced Settings Editor* option in the dropdown menu that provides more fine-grained control of JupyterLab settings and configuration options.
- **Help:** A list of JupyterLab and kernel help links.

The left sidebar in JupyterLab contains a number of commonly used tabs, such as a file browser (showing the contents of the directory where the JupyterLab server was launched), a list of running kernels and terminals, the command palette, and a list of open tabs in the main work area.

The main work area in JupyterLab enables you to arrange documents (notebooks, text files) and other activities (terminals, consoles) into panels of tabs that can be resized or subdivided. A screenshot of the default main working area is provided below.

### Key Features of JupyterLab:
- **Code cells:** These are cells where you write and run code.
- **Markdown cells:** These cells are used for explanatory text, formatted using Markdown.
- **Rich outputs:** Visualizations, graphs, images, and other media can be displayed inline within the notebook.
- **File management:** JupyterLab has a built-in file browser that allows you to organize and manage notebooks, text files, images, and other resources.

### How to Add Cells in JupyterLab
In JupyterLab, cells are the building blocks of your notebook. You can add, delete, and modify cells to create a flow between running code and explanatory text.

**Note:** The keyboard shortcuts below (B, A, Y, M and I) only work in **command mode**, which means you are not typing inside a cell. Press **Esc** first to leave the cell and make sure you are in command mode.

#### Adding Cells:

1. **Keyboard Shortcuts:** Press **B** to insert a cell below the current one or **A** to insert a cell above.

2. **Clicking Buttons:** Use the "+" button in the toolbar to insert a new cell.

#### Switching Between Cell Types (Code and Markdown):
You can switch between code cells and markdown cells to organize your content effectively.

1. **To Change a Cell Type to Code:**

- Select the cell and press **Y**.
- Alternatively, click on the cell type dropdown menu in the toolbar and select *Code*.

2. **To Change a Cell Type to Markdown:**

- Select the cell and press **M**.
- Alternatively, use the cell type dropdown menu and select *Markdown*.

This allows you to seamlessly switch between running code and writing descriptive text.

#### Running and Stopping Cells
You can run (execute) a cell to see the result of its code, or stop a cell that is still running.

1. **Running Cells:**
To run a cell, you can:

- Windows/Linux: Press **Shift + Enter** to run the current cell and move to the next one. You can also use **Ctrl + Enter** to run the cell and stay in that cell.

- Mac: Press **Shift + Return** to run the current cell and move to the next one. You can also use **Command + Return** to run the cell and stay in that cell.

- Alternatively, click the *Run* button in the toolbar.

2. **Stopping a Cell:**
If you want to stop a cell that is running (for example, if it is taking too long):

- Go to *Kernel → Interrupt Kernel* in the top menu or use the *Stop* button in the toolbar.

- Alternatively, use the keyboard shortcut **I** twice (press I quickly two times) to interrupt the current execution.

## Variables in Python
In the previous example, we saw that we can use Python as an advanced calculator. However, the results of these calculations are lost after we perform them. If we want to save a value and reuse it in our code, we need to store it in a place in the computer’s memory so we can access it later.

**Variables** are names that we give to a location in the computer’s memory. Variables are *not* the values themselves; they point to the memory location where the actual values are stored.

A few rules for naming variables in Python:
- Variable names can only contain letters, digits, and underscores `_` (typically used to separate words in long variable names).
- Variable names cannot start with a digit.
- Variable names are **case sensitive**. For example, `age`, `Age` and `AGE` are three different variables.

The name we choose for a variable should be meaningful, so that you or another programmer know what it holds.

In Python, the `=` symbol **assigns** the value on the right to the name on the left. The variable is created when a value is assigned to it. Here is how we can assign a number to a variable called `age` in Python:

````python
# Defining a variable called age and assigning a number to this variable
age = 26
age
````

You can hold **Shift** and press **Enter** to run this cell. Now the value 24 is stored in a memory location in the computer called `age`. If I want to see what is inside the variable `age`, I can use a ***built-in function*** in Python called `print()`. 

The values passed to the `print()` function go inside the parentheses, and they are called **arguments**. We can write:

````python
# Using print function to print something on screen
print(age)
````

Instead of numbers, we can also assign a series of characters, called a **string**, to a variable. Note that strings should be put inside single quotes or double quotes in Python.

````python
# Defining a variable called first_name and assigning a series of characters to this variable
first_name = "Tannia"
print(first_name)
````

We can use the `print()` function to print two or more variables with some text in between:

## Data types in Python
Every value (data) in a Python program has a specific type. There are four main types of values in Python:
- **Integer (int)**: represents positive or negative whole numbers like 4 or -12.
- **Floating point number (float)**: represents real numbers (numbers with a decimal point) like 3.14 or -2.5.
- **Character string (usually called “string”, str)**: text.
- **Boolean (bool)**: represents one of two values, `True` or `False`.

In Python, we can use the built-in function `type()` to find out what type a value has. `type()` works on variables as well, but remember that it is the *value* that has the type; the variable is just a label.


````python
print(type(14))
print(type(first_name))
print(type(3.4))
````

### Exercise 1.3
Write Python code that asks the user for their height in centimeters, converts it to feet, and prints it out.

````python
# Ask the user for their height in centimeters
height_cm = float(input("Enter your height in centimeters:"))

# Convert centimeters to feet (1 cm = 0.0328084 feet)
height_feet = height_cm * 0.0328084

# Print the result (the f before the quotes makes this an f-string, which we cover next)
print(f"Your height in feet is : {height_feet:.2f} ft")
print("Your height in feet is :", round(height_feet, 2), "ft")
````

As a final note, we will explore an alternative way to display a string or the contents of a variable in Python, called formatted string literals, or **f-strings** for short. This feature was added to Python in version 3.6. The real benefit of f-strings shows when you need to build a string from multiple values. Rather than joining several strings together, you can put the name of a variable, or even an expression, directly inside the string. This makes your code clearer and more concise, and it becomes very useful as you write more code and your programs become more complex.


f-strings are strings that have an `f` before the opening quotation mark. They can include Python expressions inside curly braces `{}`. Python replaces those expressions with their values. This is why f-strings are known as a *string interpolation* tool (a way of inserting values into a string).

````python
name  = "Tannia"
f"Hello, {name}! You're {age} years old."
````

As you can see, the code is more readable and concise using the f-string syntax. You can also control how numbers are displayed inside an f-string. For example, in Exercise 1.3 we used `{feet:.2f}`, which shows the value of `feet` with two decimal places.

## Functions in Python
A **function** is a block of code that performs a specific task and only runs when it is called. Suppose you need to write a program that creates a blank box and then fills the box with some characters. You can create three functions to solve this problem: a function that creates a blank box, a function that reads the characters from the user, and a function that fills the box with those characters. This way, we can divide a complex problem into smaller pieces, which makes our program easier to understand and reuse.

There are three different types of functions in Python:
- **Built-in functions**, which come with the base version of Python.
- **External-library functions**, which are defined inside external libraries and come with those libraries.
- **User-defined functions**, which you can define yourself.

We have already used a few built-in functions. For example, `print()` is a built-in function that prints the specified message on the screen (or another standard output device).

To use a function, we need to **call** it. Any function that is called in Python should have parentheses after its name. An argument is a value inside the parentheses that is passed into the function. `print()`, for example, can take zero or more arguments. `print()` with no arguments (zero arguments) prints a blank line.

````python
print("first line")
print()
print("second line")
````

Every function call produces a result. If the function doesn’t have a useful result to return, it usually returns the special value `None`. `None` is a Python object that stands in any time there is no value. For example, `print()` displays text on the screen, but it doesn’t return a useful result. So if we save the result of `print()` in a variable and then print that variable, we get `None`:

## Type conversions
The process of converting one Python data type into another is known as **type conversion**. There are two main types of type conversion in Python: implicit type conversion and explicit type conversion.

In an **implicit type conversion**, Python converts the data type automatically while the code is running, so we don’t have to do it ourselves. See the example below of an implicit type conversion:

````python
first_number = 4
second_number = 4.25
addition = first_number + second_number
type(addition)
````

In the example above, we took two variables, one with an integer data type and one with a float data type, and added them together. 

We stored the result of the addition in another variable named `sum`. When we checked the data type of `sum`, we saw that Python had automatically converted it into the float data type. This is called implicit type conversion. Python converted `sum` into a float, and not an integer, because converting it into an integer would have removed the fractional part (.25), which would have resulted in data loss. So, Python converts to the data type that can hold more information (from int to float) to prevent the loss of data.

**Explicit type conversion** is also known as *typecasting*. In an explicit type conversion, you tell Python exactly which type you want to convert the data into. Python has several built-in functions for explicit type conversion, but here we will focus on the main three:

- `int(x)`: converts `x` to an integer.
- `float(x)`: converts `x` to a float.
- `str(x)`: converts `x` to a string.

Let’s see what happens when we try to add an integer and a string:

````python
number = 4
string = '5.5'
addition = number + string
print(addition)
````

As expected, Python produces an error in response to our code because it cannot add an integer and a string. To fix the code, we first need to convert the string into a float and then add the two numbers.

````python
number = 4
string = '5.5'
addition = number + float(string)
print(addition)
````

Now let’s look at another situation, where we convert a number into a string and then add a few strings together. When we add strings, Python joins them exactly as they are, with **no spaces** in between:

````python
number = 26
string1 = "Years"
string2 = "Tannia"
addition = str(number) + " " + string1 + string2
addition
````

## More built-in functions
In addition to `print()`, `input()` and `type()`, there are a lot of built-in functions in Python that we can use to make our calculations easier. A few commonly used statistical built-in functions in Python are `min()`, which finds the minimum (smallest) value, and `max()`, which finds the maximum (largest) value. Note that `min()` and `max()` both work on character strings as well as numbers. To decide which character is larger or smaller, Python uses the order **0-9, then A-Z, then a-z** (so digits come before uppercase letters, and uppercase letters come before lowercase letters).

````python
print(min(1,30,20))
print(max(1,30,20))
print(min('a', 'A'))
print(max(1,'a'))
````

Note that the arguments of `min()` and `max()` must be things that can meaningfully be compared; if not, Python will return an error, as you can see in the last line below.

In addition to `min()` and `max()`, we can use the `round()` function to round a floating-point number. In the `round()` function, we can also specify the number of decimal places (`n`) we want for the value `x` by using the following format: `round(x, n)`. If we leave out `n`, the number is rounded to the nearest whole number.

````python
round(2.673838485956868699, 2)
````
