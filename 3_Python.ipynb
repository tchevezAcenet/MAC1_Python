Here we use the **Leibniz series**: this is a math pattern where, if you keep adding and subtracting the fractions 1 - 1/3 + 1/5 - 1/7 + ... and multiply the result by 4, you get something close to π (which is about 3.14159).

The more terms you include, the closer you get to the real value of π.

---

### What does `equation(k)` do?
- It calculates one small piece (term) of a special math pattern.
- For example, if you give it 0, it gives you the first piece.
- If you give it 1, it gives you the second piece, and so on.

### What does `summation(n)` do?
- It adds up the first `n` pieces from the pattern.
- It does this by:
    - If `n` is 0, it just returns 0 (because there’s nothing to add).
    - If `n` is not 0, it:
        - Finds the last piece (`equation(n-1)`).
        - Adds it to the sum of all the pieces before it (`summation(n-1)`).

- This means it **keeps calling itself** to add smaller and smaller parts until it reaches zero.
### How does this look when running?
Say you want to add the first 3 pieces (`summation(3)`):
- It finds piece number 2 (`equation(2)`), and adds it to…
- The sum of the first 2 pieces (`summation(2)`), which does the same:
    - Finds piece number 1 (`equation(1)`), plus…
    - Sum of the first 1 piece (`summation(1)`), which finds piece number 0 (`equation(0)`) plus…
    - Sum of 0 pieces (`summation(0)`) which returns 0 to stop.

Then it adds everything back up step-by-step.

### What happens in the loop?
- The loop runs the sum for different numbers of pieces: 1, 10, 100, and 1000.
- It prints out the result for each to show how the sum gets closer to the number π as you add more pieces.

### In short:
- `equation(k)` gives one piece.
- `summation(n)` adds up `n` pieces by calling itself smaller and smaller until it reaches zero.
- This is called **recursion** — solving a big problem by breaking it down into smaller pieces of the same problem.

### Note:
Python limits how deep recursion can go. The default limit is around 1000 recursive calls (Jupyter may set it a little higher). So if you try something like `summation(10000)`, the program goes too deep and crashes with a `RecursionError`.
