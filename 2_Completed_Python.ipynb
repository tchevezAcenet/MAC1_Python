## Lists in Python
**Lists** are used to store multiple items in a single variable. Lists are one of the built-in data types in Python used to store collections of data, together with tuples and dictionaries, each with different qualities and uses. If we run an experiment and collect a physical quantity like the temperature of water over time, then we may end up with multiple variables like `temperature_001`, `temperature_002`, `temperature_003`, etc. Doing calculations with a hundred variables called `temperature_001`, `temperature_002`, etc., would be at least as slow as doing them by hand. Instead, we can use a list to store many values together inside square brackets `[]`. Remember that the values should be separated by commas:

````python
temperatures = [18,19,20,17.5,22]
print("Temperatures are:", temperatures)
````

We can also create a list using the `list()` constructor (note the double parentheses):

````python
temperatures = list((18,19,20,17.5,22))
print("Temperatures are:", temperatures)
````

It is also important to know how to initialize (create) empty lists. In many situations, you can solve problems in data engineering by using an empty list, for example as a placeholder that will later be filled in with data.

````python
empty_list = []
print(empty_list)
````

We can use the built-in function `len()` to find the length of a list, or how many values are in a list.

````python
print('length of Temperatures:', len(temperatures))
````

List items are **indexed**: the first item has index `[0]`, the second item has index `[1]`, and so on. The last item has index `[-1]`.

````python
print("Item in index 0:", temperatures[0])
print("Item in index 3:", temperatures[3])
print("Item in index -1:", temperatures[-1])
````

````python
print(temperatures[len(temperatures)-1])
````

A list can contain different data types:

````python
list_example = ["abc", 34, True, 40.2, 'xyz', 40.2]
list_example
````

List items are ordered, **changeable**, and allow duplicate values. To replace a value, use an index expression on the left side of the assignment:

````python
temperatures[0] = 14.8
print("Temperatures is now:", temperatures)
````

To access a range of elements in a list, you need to **slice** the list. One way to do this is to use the simple slicing operator (`:`). With this operator, you can specify where to start the slicing, where to end it, and (optionally) the step. Note that the slice **includes the start index but stops just before the end index**, so `List_alphabets[2:7]` gives the items at indices 2, 3, 4, 5 and 6. List slicing returns a new list from the existing list.

````python
list_alphabets = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i']
subset_alphabet = list_alphabets[2:7]
subset_alphabet
````

If you add new items to a list, the new items will be placed at the end of the list. Use `list_name.append()` to add items to the end of a list.

````python
primes = [2,3,5]
print("Primes was initially: ", primes)
primes.append(7)
print("Primes has become:", primes)
````

Here, `append()` is a **method** of lists. A method is like a function, but it is tied to a particular object. Use `object_name.method_name()` to call methods. This deliberately resembles the way we refer to things in a library. We will meet other methods of lists as we go along. Use `help(list)` for a preview.

`extend()` is similar to `append()`, but it allows you to combine two lists.

````python
teen_primes = [11,13,17,19]
middle_aged_primes = [37,41,43,47]
print("Primes is currently:", primes)
````

````python
primes.extend(teen_primes)
print("Primes has now become:", primes)
````

````python
primes.extend(middle_aged_primes)
print("Primes final version is:", primes)
````

Note that while `extend()` keeps the list flat, appending a list to a list means the last element in `primes` will itself be a list, not an integer. Lists can contain values of any type; therefore, lists of lists are possible.

The `count()` method returns the number of elements with the specified value.

````python
fruits = ['apple', 'cherry', 'banana', 'cherry']
number_of_cherries = fruits.count("cherry")
print(number_of_cherries)
````

The characters (individual letters, numbers, and so on) in a string are ordered. For example, the string `'AB'` is not the same as `'BA'`. Because of this ordering, we can treat a string as a list of characters. Each position in the string (first, second, etc.) is given a number, called an index. Therefore, character strings can be indexed like lists, and indices are **numbered from 0**. We can use the position’s index in square brackets to get the character at that position. Because counting starts at 0, `element[0]` is the first character (sometimes called the “zeroth” character), and `element[3]` is actually the *fourth* character.

````python
element = 'carbon'
print('First Character:', element[0])
print('Character at index 3:', element[3])
````

Note that character strings are **immutable** (unchangeable): you cannot change the characters in a string after it has been created. In contrast, lists are **mutable**, which means they can be modified in place. Python considers a string to be a single value with parts, not a collection of values.

````python
element[0] = 'C'
````

Python reports an `IndexError` if we attempt to access a value that doesn’t exist.

````python
element[99]
````

A part of a string is called a **substring**. A substring can be as short as a single character. An item in a list is called an **element**. Whenever we treat a string as if it were a list, the string’s elements are its individual characters. A **slice** is a part of a string (or, more generally, a part of any list-like thing). We take a slice with the notation `[start:stop]`, where `start` is the index of the first element we want and `stop` is the index of the element just after the last element we want. The difference between `stop` and `start` is the slice’s length. Taking a slice does *not* change the contents of the original string. Instead, it returns a copy of part of the original string.


````python
atom_name = 'sodium'
print(atom_name[0:3])
print(atom_name[:3])
````

## Tuples
**Tuples** are used to store multiple items in a single variable. A tuple is one of the main built-in data types in Python used to store collections of data, like a list. A tuple is a collection which is **ordered and unchangeable**. Like list items, tuple items are indexed: the first item has index `[0]`, the second item has index `[1]`, etc. Since tuples are indexed, they can have items with the same value. Tuples are also ordered, which means that the items have a defined order and that order will not change. 

````python
tuple_cars = ('toyota', 'honda', 'ferrari')
print("My first tuple:", tuple_cars)
print("Second element of my tuple:", tuple_cars[1])
````

Unlike lists, tuples are unchangeable which means that we cannot change, add or remove items after the tuple has been created.

````python
tuple_cars[1] = 'lamborghini'
````

**Note:**
- Tuples are written with round brackets `()`.
- Tuples are ordered.
- Unlike lists, tuple items are unchangeable (immutable).

Just like with other Python data types such as lists and strings, we can find out how many items are in a tuple using the `len()` function.


````python
len(tuple_cars)
````

A tuple can contain different data types.

````python
tuple_stuff = ('toyota', 5.1, [3,1,0])
tuple_stuff[-1]
````

Tuples have two built-in methods: `count()`, which returns the number of times a value appears in the tuple, and `index()`, which returns the position of the first time a value appears.

````python
tuple_numbers = (5,3,5,2,1,-1,3)
print("Occurences of 5:", tuple_numbers.count(5))
print("Index of 3:", tuple_numbers.index(3))
````

## Dictionaries
A **dictionary** in Python is a collection of **key-value pairs**, similar to how a real dictionary has words (keys) and their meanings (values). Each key in a dictionary must be **unique**, and you can use it to access its corresponding value. Dictionaries are defined using curly braces `{}`.

````python
student_data = {
    "name" : "Alex Johnson",
    "age" : 20,
    "grade" : 88.5,
    "major" : "Computer Science"
}
student_data
````

In this example, we have a dictionary called `student_data` that stores various pieces of information about a student's academic details.

#### Using Dictionaries:

**1. Accessing Values:** You can retrieve a value using its corresponding key.

````python
current_grade = student_data["grade"]
current_grade
````

**2. Adding or Modifying Values:** You can easily add new key-value pairs or modify existing ones:

````python
# Modify an existing value
student_data["grade"] = 92.0
student_data
````

````python
# Add a new key-value pair
student_data["attendance"] = 95
student_data
````

**3. Removing Key-Value Pairs:** You can remove a key-value pair using the `del` statement:

````python
del student_data["major"]
student_data
````

Now, let’s expand this dictionary to include more details by combining dictionaries and lists. We can represent multiple subjects with their scores and attendance records, offering a detailed view of the student’s academic performance.

````python
student_data_extension = {
    "name": "Alex Johnson",
    "subjects" : [
        {"subject" : "Math", "score" : 85.0, "attendance": 90},
        {"subject" : "English", "score" : 88.5, "attendance": 95},
        {"subject" : "Science", "score" : 92.0, "attendance": 85}
    ]
}
````

The `student_data_extension` example is a well-organized way to store information about a student's academic performance using a dictionary with a list of dictionaries inside. The main dictionary contains keys for the student's name and a list of subject records (each with details like score and attendance).

This structure is beneficial because it keeps related information together, making it easy to find and update data. The use of a list allows for multiple subjects to be stored in order, while each subject's details are kept in their own dictionary.

This combination of lists and dictionaries makes it simple to manage complex datasets in education and reflects how we naturally group information in real life.

**Accessing Top-Level Data**

**1. Access the Name:**
To get the name of the student from the data, you can use the key `"name"`:

````python
name = student_data_extension["name"]
name
````

**Accessing Data from the Subjects List**

**2. Access the Subjects:**
The `"subjects"` key points to a list of subject dictionaries. To access this list, you would write:

````python
subjects = student_data_extension["subjects"]
subjects
````

**3. Access Individual Subjects:**

Each subject in the list can be accessed using its index (position in the list). Remember that Python uses **zero-based indexing**.

For example, to access the first subject:

````python
first_subject = student_data_extension["subjects"][0]
first_subject
````

````python
subjects[0]
````

**4. Access Specific Data in a Subject:**

You can also directly access specific details within a subject. For example, to get the score of the second subject:

````python
second_subject_score = student_data_extension["subjects"][1]["score"]
second_subject_score
````

````python
subjects[1]["score"]
````

**5. Access Score and Attendance:**

Similarly, you can access the score and attendance of any subject. For example, to get the attendance of the third subject:

````python
third_subject_attendance = student_data_extension["subjects"][2]["attendance"]
third_subject_attendance
````

````python
subjects[2]["attendance"]
````

````python
third_subject_score = student_data_extension["subjects"][2]["score"]
third_subject_score
````

````python
subjects[2]["score"]
````

To access data in the `student_data_extension` dictionary, you use the keys to reach the specific pieces of information you need. You can start with top-level keys like `"name"` and then drill down into lists and dictionaries, using indices and additional keys to get to the exact data point you’re interested in. This method allows for easy retrieval and manipulation of structured data.
